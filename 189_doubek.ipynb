```python
import os
import polars as pl
from src.bez_bordelu import bez_bordelu
from src.najdi_rok import najdi_rok
from src.pocet_stran import pocet_stran
pl.Config(tbl_rows=500)
```

```python
df = pl.read_parquet(
    os.path.join("data","cnb_vyber.parquet")
).with_columns(
    pl.col('300_a').map_elements(pocet_stran, return_dtype=int).alias('stran')
).with_columns(
    pl.col('008').map_elements(najdi_rok, return_dtype=int).alias('rok')
).with_columns(
    pl.col('245_a').map_elements(bez_bordelu, return_dtype=str)
)
```

```python
df.filter(pl.col("245_a").str.contains("Co je uvnitř"))
```

```python
df.explode("700_7").filter(pl.col("700_7") == "jk01102794").select(pl.col(['rok','100_a','245_a'])).sort(by='rok')
```
